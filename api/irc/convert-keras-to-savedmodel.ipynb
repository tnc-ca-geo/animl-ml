# Convert Keras Model to TensorFlow SavedModel

This notebook loads the IRC_2.h5 Keras model, converts it to TensorFlow SavedModel format, and zips the model artifacts into a .tar.gz file.

```python
import tensorflow as tf
import tensorflow_addons as tfa
# Check TensorFlow version
print(tf.__version__)
```

```python
ORIGINAL_MODEL_PATH = "./original-model/IRC_2.h5"
```

```python
# Load the Keras .h5 model
# NOTE: if adapting this code for converting non-mewc trained models in the future, you may not need the custom_objects argument
model = tf.keras.models.load_model(ORIGINAL_MODEL_PATH, custom_objects={'SigmoidFocalCrossEntropy': tfa.losses.SigmoidFocalCrossEntropy})
print('Model loaded successfully!')
```

```python
# Export the model to TensorFlow SavedModel format

# Note: This directory structure will need to be followed - see notes for the next section
model_version = '2'
export_root = 'tf-saved-model/'
export_dir = export_root + model_version


tf.saved_model.save(model, export_dir)
print(f'SavedModel exported to {export_dir}')
```

```python
# create tarfile
import tarfile

with tarfile.open('tf-saved-model.tar.gz', mode='w:gz') as archive:
    archive.add(export_root, recursive=True)
print(f'Model artifacts zipped to tf-saved-model.tar.gz')
```
