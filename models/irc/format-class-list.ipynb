# Format class list
The class_list.yaml is a list of human-readable class names next to integers (out-of-order Ids, maybe?), but it appears that their index in that list corresponds to the order of the predictions array we're receiving from the model, so we're converting the .yaml list into a index_to_name.json map, using their index in the original .yaml file as the keys/Ids. I am a little uncleaer if this class_list.yaml structure is specific to the mewc-train output or AddaxAI. The original model files were copied from AddaxAI, so they may have been modified to integrate wit AddaxAI.

NOTE: this step is specific to this model, so you may need to modify this notebook if you are adapting this workflow to deploy a different Keras/Tensorflow model to Animl. The rest of the code does expect that there is an `exported-model/index_to_name.json` file that is structured as follows, so adjust accordingly:

```javascript
{"0": "Human", "1": "Non-Living Object", "2": "bird sp", ...}
```

````python
import yaml
import json
````

````python
PATH_TO_CLASS_LIST = "./original-model/class_list.yaml"
OUTPUT_FILE = "./exported-model/index_to_name.json"
````

````python
def format_class_list(yaml_file):
    with open(yaml_file, 'r') as f:
        class_list = yaml.safe_load(f)
    
    # convert to the index_to_name dict structure
    index_to_name = {}
    for i, name in enumerate(class_list):
        index_to_name[f"{i}"] = f"{name}"
    return index_to_name

def save_formatted_class_list(class_list, output_file):
    with open(output_file, 'w') as f:
        json.dump(class_list, f)
    print(f"Formatted class list saved to {output_file}")
````

````python
class_list = format_class_list(PATH_TO_CLASS_LIST)
print(class_list)

save_formatted_class_list(class_list, OUTPUT_FILE)
````
